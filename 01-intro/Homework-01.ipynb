```python
subset_df = asian_cars[['vehicle_weight', 'model_year']]
```

```python
x=subset_df.head(7).values
x
```

```python
xtx = x.T.dot(x)
xtx
```

```python
inv_xtx = np.linalg.inv(xtx)
inv_xtx
```

```python
y=np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```
